### A1 · The invoice  `[5 marks]`  ·  **★ Foundation**

A shop sells a product. These facts arrive as **text** from a form:

| `gst_percent` | `18` |

Create these variables, convert the text values to numbers, and calculate: `subtotal` (price × quantity), the discount amount, the amount after discount, the GST (on the amount after discount) and the `grand_total`.

Print the invoice in **exactly** this format (money with commas and two decimals; use f-strings, and let Python do all calculations):

````python
# Write your answer here
item = "Wireless Mouse"
price = float("749.50")
qty = int("3")
discount_percent = 10
gst_percent = 18
subtotal = price * qty
discount = subtotal * (discount_percent/ 100)
amount_after_discount = subtotal - discount
gst = amount_after_discount * (gst_percent / 100)
grand_total = amount_after_discount + gst
print('----- INVOICE -----')
print(f'{'Item':<12}: {item} x {qty}')
print(f'{'Subtotal':<12}: Rs {subtotal:.2f}')
print(f'{'Discount'} ({discount_percent}%): Rs {discount:.2f}')
print(f'{'After disc.':<12}: Rs {amount_after_discount:.2f}')
print(f'GST ({gst_percent}%) {':':>4} Rs {gst:.2f}')
print(f'{'Grand total':<12}: Rs {grand_total:.2f}')
````

### A4 · Printing a neat table  `[6 marks]`  ·  **★★ Standard**

A shop prints its bills in a fixed-width table. The items are in the given-data cell below. Use **f-string alignment** to print **exactly** the layout shown under *Expected output*.

Alignment in an f-string: `{text:<12}` left-aligns in 12 characters, `{text:>12}` right-aligns, `{text:^12}` centres, and a character before the symbol is the fill (`{text:*^12}`). Numbers can also carry a format: `{value:>10.2f}`.

Column widths: **Item** 16 (left) · **Qty** 5 (right) · **Price** 10 (right, 2 decimals) · **Amount** 12 (right, commas and 2 decimals). Amount = qty × price.

1. The title ` ITEM BILL ` (a space on each side) **centred in 43 characters** and padded with `=`.
2. The column headings: `Item` left-aligned, the other three right-aligned, with the widths above.
3. A line of **43** dashes.
4. One row for every item.
5. A line of 43 dashes, then a `Total` row: the word `Total` left-aligned in **31** characters followed by the total amount right-aligned in 12 characters with commas and 2 decimals.

````python
# 📦 GIVEN DATA: run this cell, do not edit it
items = [("Notebook", 3, 45.5), ("Pen", 12, 8.25), ("Wireless Mouse", 1, 749.5), ("USB Cable", 2, 199)]
````

````python
# Write your answer here
print(f'{'='*16} ITEM BILL {'='*16} ')
print(f'{'Item':<16}{'Qty':>5}{'Price':>10}{'Amount':>12}')
print(f'{'-'*43}')
total=0
for i in items:
  print(f'{i[0]:<15} {i[1]:>5}{i[2]:>10.2f}{i[1]*i[2]:>12.2f}')
  total+=i[1]*i[2]
print(f'{'-'*43}')
print(f'{'Total':<31}{total:>12,.2f}')
````

**Expected output:**

```text
================ ITEM BILL ================
Item              Qty     Price      Amount
-------------------------------------------
Notebook            3     45.50      136.50
Pen                12      8.25       99.00
Wireless Mouse      1    749.50      749.50
USB Cable           2    199.00      398.00
-------------------------------------------
Total                              1,383.00
```
