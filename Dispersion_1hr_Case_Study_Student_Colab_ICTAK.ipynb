## The scenario
Two classes both have an average mark of **50**. But in Class A everyone scores close to 50, and in Class B the marks are all over the place. The principal asks you: *which class is more consistent?* Central tendency (mean) cannot answer this. **Dispersion** can.

## What you will practise
* **Range** = maximum − minimum
* **Variance** = average of the squared differences from the mean
* **Standard deviation** = square root of the variance
* Population vs sample (`ddof`)
* Comparing groups with a function and with pandas `groupby`

````python
import numpy as np
import pandas as pd

class_a = [48, 49, 50, 51, 52]
class_b = [20, 35, 50, 65, 80]
sample  = [2, 4, 6, 8, 10]          # the worked example from the class notes

sections = pd.DataFrame({
    "Section": ["A"] * 6 + ["B"] * 6,
    "Marks":   [62, 65, 60, 63, 64, 61, 30, 95, 55, 80, 45, 90],
})
print("Data ready")
````

### Task 3 · Variance step by step  ·  *Beginner* (5 marks)
Follow the 4 steps from the notes for `sample = [2, 4, 6, 8, 10]`:
1. mean
2. difference of each value from the mean
3. square each difference
4. divide the sum of squares by the number of values

Print each step.

**Expected output**
```
Task 3: Variance step by step
Mean: 6.0
Differences: [-4.0, -2.0, 0.0, 2.0, 4.0]
Squared differences: [16.0, 4.0, 0.0, 4.0, 16.0]
Sum of squares: 40.0
Variance: 8.0
```

````python
# Write your code here
mean_val = np.mean(sample)
diffs = [float(x - mean_val) for x in sample]
sq_diffs = [d ** 2 for d in diffs]
sum_sq = sum(sq_diffs)
var_val = sum_sq / len(sample)

print("Task 3: Variance step by step")
print(f"Mean: {mean_val}")
print(f"Differences: {diffs}")
print(f"Squared differences: {sq_diffs}")
print(f"Sum of squares: {sum_sq}")
print(f"Variance: {var_val}")
````

### Task 4 · Variance with NumPy  ·  *Beginner* (5 marks)
Use `np.var()` to find the variance of `sample`, `class_a` and `class_b`. (Check that `sample` gives the same answer as Task 3.)

### Task 5 · Standard deviation  ·  *Beginner* (5 marks)
**Standard deviation = square root of variance.** Use `np.std()` for Class A and Class B and print them rounded to 2 decimals. Then check one of them with `np.sqrt(np.var(...))`.

**Expected output**
```
Task 5: Standard deviation
Class A std: 1.41
Class B std: 21.21
Check A: 1.41
```

````python
# Write your code here
std_a = round(np.std(class_a), 2)
std_b = round(np.std(class_b), 2)
check_a = round(np.sqrt(np.var(class_a)), 2)

print("Task 5: Standard deviation")
print(f"Class A std: {std_a}")
print(f"Class B std: {std_b}")
print(f"Check A: {check_a}")
````

### Task 6 · Which class is more consistent?  ·  *Beginner* (5 marks)
Compare the standard deviation of the two classes with an `if / else` and print which class is **more consistent** (smaller spread).

**Expected output**
```

```

````python
# Write your code here
print("Task 6: Consistency")
if np.std(class_a) < np.std(class_b):
    print("Class A is more consistent")
else:
    print("Class B is more consistent")
````

### Task 7 · Dispersion report function  ·  *Intermediate* (5 marks)
Write a function `dispersion_report(data)` that **returns** a dictionary with the keys `range`, `variance` and `std` (variance and std rounded to 2 decimals). Call it for Class A and Class B and print the results.

**Expected output**
```
Task 7: Report
Class A: {'range': 4, 'variance': 2.0, 'std': 1.41}
Class B: {'range': 60, 'variance': 450.0, 'std': 21.21}
```

````python
# Write your code here
def dispersion_report(data):
    return {
        'range': max(data) - min(data),
        'variance': float(np.var(data)),
        'std': round(float(np.std(data)), 2)
    }

print("Task 7: Report")
print(f"Class A: {dispersion_report(class_a)}")
print(f"Class B: {dispersion_report(class_b)}")
````

### Task 8 · Population vs sample  ·  *Intermediate* (5 marks)
`np.var()` divides by **n** (population variance). For a sample we divide by **n − 1**, which in NumPy is `ddof=1`. Print both variances of `sample`.

**Expected output**
```
Task 8: Population vs sample
Population variance: 8.0
Sample variance: 10.0
```

````python
# Write your code here
print("Task 8: Population vs sample")
print(f"Population variance: {np.var(sample)}")
print(f"Sample variance: {np.var(sample, ddof=1)}")
````

### Task 9 · Spread of two sections (pandas)  ·  *Intermediate* (5 marks)
In the `sections` table, use `groupby("Section")` on `Marks` with `.agg(["mean", "min", "max", "std"])`, rounded to 2 decimals. Which section is more spread out? *(pandas `std` uses n − 1.)*

**Expected output**
```
Task 9: Sections
          mean  min  max    std
Section                        
A        62.50   60   65   1.87
B        65.83   30   95  26.35
```

````python
# Write your code here
print("Task 9: Sections")
print(sections.groupby("Section")["Marks"].agg(["mean", "min", "max", "std"]).round(2))
````
