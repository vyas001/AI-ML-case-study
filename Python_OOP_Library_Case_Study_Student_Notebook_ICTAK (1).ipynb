## Step 1 · The `Book` class
| Class / method | What it does |
|---|---|
| Book(book_id, title, author, copies) | stores the details; the number of copies is **private** (`self.__copies`) |
| Book.available_copies() | returns the number of copies left |
| Book.issue() | if a copy is left, reduce the copies by one and return `True`; otherwise return `False` |
| Book.give_back() | adds one copy back |
| Book.__str__() | returns `B101 | Python Basics by Anu | Copies: 2` |

> Use `self.__copies` (two underscores) so that the number of copies is **private**.

````python
# Write the Book class here
class Book:
    def __init__(self,book_id,title,author,copies):
        self.book_id=book_id
        self.title=title
        self.author=author
        self.__copies=copies
    def available_copies(self):
        return self.__copies
    def issue(self):
        if(self.__copies>0):
            self.__copies-=1
            return True
        else:
            return False
    def give_back(self):
        self.__copies+=1
    def __str__(self):
        return f'{self.book_id} | {self.title} by {self.author} | Copies: {self.__copies}'
        
````

## Step 2 · `Member`, `Student` and `Teacher`
| Class / method | What it does |
|---|---|
| Member(member_id, name) | has an empty list `borrowed`; `max_books()` returns 1, `member_type()` returns `"Member"`; `can_borrow()` is `True` while the borrowed list is shorter than `max_books()`; `__str__()` returns `M1 | Anu | Student | Borrowed: 0/2` |
| Student(Member) | **inherits** from `Member`; overrides `max_books()` (2 books) and `member_type()` (`"Student"`) |
| Teacher(Member) | inherits from `Member`; overrides `max_books()` (4 books) and `member_type()` (`"Teacher"`) |

> Write `Member` first. `Student` and `Teacher` should be very short: they **inherit** everything and only **override** two methods.

````python
# Write the Member class here
class Member:
    def __init__(self,member_id,name):
        self.member_id=member_id
        self.name=name
        self.borrowed=[]
    def max_books(self):
        return 1
    def member_type(self):
        return 'Member'
    def can_borrow(self):
        if (len(self.borrowed)<self.max_books()):
            return True
        else:
            return False
    def __str__(self):
        return f'{self.member_id} | {self.name} | {self.member_type()} | Borrowed: {len(self.borrowed)}/{self.max_books()}'
    
````

````python
# Write the Student and Teacher classes here
class Student(Member):
    def __init__(self, member_id, name):
        super().__init__(member_id, name)
    def max_books(self):
        return 2
    def member_type(self):
        return 'Student'

class Teacher(Member):
    def __init__(self, member_id, name):
        super().__init__(member_id, name)
    def max_books(self):
        return 4
    def member_type(self):
        return 'Teacher'
````

## Step 3 · The `Library` class
| Class / method | What it does |
|---|---|
| Library() | holds two dictionaries: `books` (id to `Book` object) and `members` (id to member object) |
| Library.load_books(data) / load_members(data) | create the objects from the starter data (a member of type `"teacher"` becomes a `Teacher`, otherwise a `Student`) |
| Library.issue_book(member_id, book_id) | returns a message: `Member not found`, `Book not found`, `Anu has reached the limit`, `Data Science 101 is not available` or, on success, `Anu issued Python Basics` (and records the id in the member's `borrowed` list) |
| Library.return_book(member_id, book_id) | returns `Anu returned Data Science 101` and gives the copy back; returns `Cannot return` if that member did not borrow that book |
| Library.show_books() / show_members() | print every book / member using `print(object)` |

> `issue_book` checks, in this order: member exists, book exists, member is under the limit, a copy is available.

````python
# Write the Library class here
class Library:
    def __init__(self):
        self.books={}
        self.members={}
    def load_books(self,data):
        for bookid, details in data.items():
            self.books[bookid]=Book(
                bookid, 
                details["title"], 
                details["author"], 
                details["copies"]
            )

    def load_members(self, data):
        for m in data:
            member_id = m["id"]
            name = m["name"]
            m_type = m["type"]
            
            if m_type.lower() == "teacher":
                self.members[member_id] = Teacher(member_id, name)
            else:
                self.members[member_id] = Student(member_id, name)

    def issue_book(self,member_id,book_id):
        if member_id not in self.members:
            return "Member not found"
        
        if book_id not in self.books:
            return "Book not found"
        
        member = self.members[member_id]
        book = self.books[book_id]
        
        if not member.can_borrow():
            return f"{member.name} has reached the limit"
        
        if book.available_copies() <= 0:
            return f"{book.title} is not available"
        
        book.issue()
        member.borrowed.append(book_id)
        return f"{member.name} issued {book.title}"

    def return_book(self, member_id, book_id):
        if member_id not in self.members or book_id not in self.books:
            return "Cannot return"
        
        member = self.members[member_id]
        book = self.books[book_id]
        
        if book_id in member.borrowed:
            member.borrowed.remove(book_id)
            book.give_back()
            return f"{member.name} returned {book.title}"
        else:
            return "Cannot return"

    def show_books(self):
        for book in self.books.values():
            print(book)

    def show_members(self):
        for member in self.members.values():
            print(member)

   

        

````
